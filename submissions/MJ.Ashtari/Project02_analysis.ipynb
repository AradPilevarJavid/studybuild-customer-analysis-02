```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
from sklearn.preprocessing import MinMaxScaler

cleaned_dataset = pd.read_excel('cleaned_dataset.xlsx')
print("Dataset Info:")
print(cleaned_dataset.info())
print("\nFirst 5 rows:")
print(cleaned_dataset.head())

# ==============================
# QUESTION 1 – Overall Customer Health
# ==============================
print("\n" + "="*50)
print("QUESTION 1: OVERALL CUSTOMER HEALTH")
print("="*50)

total_customers = cleaned_dataset['customer_id'].nunique()
total_purchases = cleaned_dataset['purchase_count'].sum()
total_revenue = cleaned_dataset['total_spending'].sum()
avg_spending_per_customer = cleaned_dataset['total_spending'].mean()
if total_purchases > 0:
    avg_order_value = total_revenue / total_purchases
else:
    avg_order_value = 0
avg_satisfaction = cleaned_dataset['satisfaction_score'].mean()
total_returned = cleaned_dataset['returned_items'].sum()
if total_purchases > 0:
    return_rate = total_returned / total_purchases
else:
    return_rate = 0

print(f"Total Customers: {total_customers}")
print(f"Total Purchases: {total_purchases}")
print(f"Total Revenue: {total_revenue:,.2f}")
print(f"Average Spending per Customer: {avg_spending_per_customer:,.2f}")
print(f"Average Order Value: {avg_order_value:,.2f}")
print(f"Average Satisfaction Score: {avg_satisfaction:.2f}")
print(f"Return Rate: {return_rate:.2%}")

# Summary
print("\n--- Summary ---")
print("""
فروشگاه تقریباً ۶۰ مشتری فعال با درآمد کل بیش از ۲۰۰,۰۰۰ واحد دارد.
میانگین هزینه‌ی هر مشتری حدود ۳,۳۰۰ واحد و میانگین ارزش هر سفارش حدود ۵۱۰ واحد است.
رضایت مشتریان در حد متوسط است (۳.۱ از ۵). با این حال، نرخ بالای بازگشت کالا (نزدیک به ۱۹٪)
نشان‌دهنده‌ی مشکلات احتمالی در کیفیت محصول یا انتظارات مشتریان است که نیاز به توجه دارند.
""")

# ==============================
# QUESTION 2 – Best City for Advertising
# ==============================
print("\n" + "="*50)
print("QUESTION 2: BEST CITY FOR ADVERTISING")
print("="*50)

customer_count = cleaned_dataset.groupby('city')['customer_id'].count()
total_revenue = cleaned_dataset.groupby('city')['total_spending'].sum()
avg_spending = cleaned_dataset.groupby('city')['total_spending'].mean()
avg_satisfaction = cleaned_dataset.groupby('city')['satisfaction_score'].mean()
total_purchases = cleaned_dataset.groupby('city')['purchase_count'].sum()
total_returned = cleaned_dataset.groupby('city')['returned_items'].sum()

city_analysis = pd.DataFrame({
    'customer_count': customer_count,
    'total_revenue': total_revenue,
    'avg_spending': avg_spending,
    'avg_satisfaction': avg_satisfaction,
    'total_purchases': total_purchases,
    'total_returned': total_returned
})

city_analysis['return_rate'] = city_analysis['total_returned'] / city_analysis['total_purchases']
city_analysis = city_analysis.sort_values('total_revenue', ascending=False)

# Show the table
print("\nCity Performance Summary:")
print(city_analysis[['customer_count', 'total_revenue', 'avg_spending', 'avg_satisfaction', 'return_rate']])

# Chart 1: Revenue by City
plt.figure(figsize=(10,6))
sns.barplot(x=city_analysis.index, y='total_revenue', data=city_analysis)
plt.title('Total Revenue by City', fontsize=14)
plt.xlabel('City')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Selected City and Rationale
selected_city = city_analysis.index[0]
print(f"\n--- Selected City: {selected_city} ---")
print("""
دلایل انتخاب:
- بالاترین درآمد کل در میان تمام شهرها.
- تعداد مشتریان خوب و رضایت بالاتر از میانگین (۳.۲).
- نرخ بازگشت کالا نسبت به سایر شهرهای برتر در حد متوسط.

محدودیت‌ها / ریسک‌ها:
- هزینه‌های تبلیغات در این شهر ممکن است بیشتر از شهرهای کوچک‌تر باشد.
- رقابت بالا ممکن است اثربخشی کمپین را کاهش دهد.
- نرخ بازگشت کالا، اگرچه در حد متوسط است، اما همچنان نیاز به پایش دارد تا ROI (بازگشت سرمایه) تضمین شود.
""")

# ==============================
# QUESTION 3 – Top 10 Customers for Loyalty Campaign
# ==============================
print("\n" + "="*50)
print("QUESTION 3: TOP 10 CUSTOMERS FOR LOYALTY CAMPAIGN")
print("="*50)

# Filter active customers (at least 1 purchase)
cleaned_dataset_active = cleaned_dataset[cleaned_dataset['purchase_count'] > 0].copy()

# Calculate recency score (lower last_purchase_days is better)
cleaned_dataset_active['recency_score'] = 1 / (cleaned_dataset_active['last_purchase_days'] + 3)

# Features for scoring
features = ['total_spending', 'purchase_count', 'satisfaction_score',
            'membership_level_numeric', 'recency_score']

df_scaled = cleaned_dataset_active[features].copy()

for col in features:
    min_val = df_scaled[col].min()
    max_val = df_scaled[col].max()

    if max_val > min_val:
        df_scaled[col] = (df_scaled[col] - min_val) / (max_val - min_val)
    else:
        df_scaled[col] = 0

scaled_features = df_scaled.values

weight_spending = 0.3
weight_frequency = 0.2
weight_satisfaction = 0.1
weight_membership = 0.1
weight_recency = 0.3

cleaned_dataset_active['loyalty_score'] = (
    df_scaled['total_spending'] * weight_spending +
    df_scaled['purchase_count'] * weight_frequency +
    df_scaled['satisfaction_score'] * weight_satisfaction +
    df_scaled['membership_level_numeric'] * weight_membership +
    df_scaled['recency_score'] * weight_recency
)

top10 = cleaned_dataset_active.sort_values('loyalty_score', ascending=False).head(10)

top10_output = top10[[
    'customer_id', 'first_name', 'city', 'membership_tier',
    'purchase_count', 'total_spending', 'last_purchase_days',
    'satisfaction_score'
]]

top10_output['Reason_for_Selection'] = 'High combined score based on spending, purchase frequency, recency, satisfaction, and membership tier.'

print("\nTop 10 Customers for Loyalty Campaign:")
print(top10_output.to_string(index=False))

print("\n--- Selection Rationale ---")
print("""
این مشتریان بیشترین پتانسیل را برای ارزش‌آفرینی بلندمدت دارند.
آنها به‌طور سخاوتمندانه هزینه می‌کنند، مرتباً خرید می‌کنند، به‌تازگی با فروشگاه تعامل داشته‌اند،
رضایت بالایی را گزارش می‌دهند و دارای سطوح عضویت برتر هستند.
پاداش دادن به آنها احتمالاً باعث افزایش وفاداری و توصیه به دیگران می‌شود.
""")

# ==============================
# QUESTION 4 – At-Risk Customers
# ==============================
print("\n" + "="*50)
print("QUESTION 4: IDENTIFY AT-RISK CUSTOMERS")
print("="*50)

# Definition: Total spending >= median AND last purchase > 180 days
median_spending = cleaned_dataset['total_spending'].median()
at_risk = cleaned_dataset[(cleaned_dataset['total_spending'] >= median_spending) &
             (cleaned_dataset['last_purchase_days'] > 180) &
             (cleaned_dataset['purchase_count'] > 0)]

print(f"""
تعریف مشتری در معرض ریسک:
- مشتریانی که هزینه‌ی کل آنها برابر یا بالاتر از سطح میانه است.
- هیچ فعالیت خریدی در ۱۸۰ روز گذشته ندارند.

تعداد کل مشتریان در معرض ریسک شناسایی‌شده: {len(at_risk)}
""")
at_risk_table = at_risk[['customer_id', 'first_name', 'city', 'total_spending',
                         'last_purchase_days', 'purchase_count', 'satisfaction_score']]
print("\nAt-Risk Customer List:")
#print(at_risk_table.to_string(index=False))
display(at_risk_table)


print("\n--- Win-Back Recommendation ---")
print("""
۱. ارسال ایمیل‌های شخصی‌سازی‌شده با تخفیف محدود به زمان (مثلاً ۲۰٪ تخفیف) در دسته‌بندی‌های مورد علاقه‌ی آنها.
۲. ارسال پیامک یادآوری یک هفته بعد.
۳. انجام یک نظرسنجی کوتاه برای درک دلیل توقف خرید آنها (مثلاً قیمت، کیفیت، یا خدمات).
""")

# ==============================
# QUESTION 5 – Impact of Discounts
# ==============================
print("\n" + "="*50)
print("QUESTION 5: DISCOUNT IMPACT ANALYSIS")
print("="*50)

discount_yes = cleaned_dataset[cleaned_dataset['discount_used'] == 'Yes']
discount_no = cleaned_dataset[cleaned_dataset['discount_used'] == 'No']

# Calculate
discount_yes_purchases = discount_yes['purchase_count'].sum()
discount_no_purchases = discount_no['purchase_count'].sum()

def calculate_metrics(df, total_purchases):
    return [
        len(df),
        df['total_spending'].mean(),
        df['purchase_count'].mean(),
        df['satisfaction_score'].mean(),
        df['returned_items'].sum() / total_purchases if total_purchases > 0 else 0
    ]

comparison_df = pd.DataFrame({
    'Metric': ['Customer Count', 'Avg Spending per Customer', 'Avg Purchase Count',
               'Avg Satisfaction', 'Return Rate'],
    'With Discount': calculate_metrics(discount_yes, discount_yes_purchases),
    'Without Discount': calculate_metrics(discount_no, discount_no_purchases)
})

display(comparison_df)

print("\n--- Conclusion ---")
print("""
مشتریانی که از تخفیف استفاده کرده‌اند، میانگین هزینه‌ی بالاتری، خریدهای بیشتری و رضایت بهتری دارند.
با این حال، این ثابت نمی‌کند که تخفیف‌ها باعث فروش بالاتر می‌شوند. به همان اندازه محتمل است
که فروشگاه به‌طور انتخابی تخفیف را به مشتریان باارزش ارائه می‌دهد، یا اینکه
مشتریانی که به دنبال تخفیف هستند ذاتاً به قیمت حساس‌ترند. برای مشخص کردن رابطه‌ی علت و معلولی،
نیاز به یک تست A/B کنترل‌شده است.
""")

# ==============================
# QUESTION 6 – Customer Segmentation
# ==============================
print("\n" + "="*50)
print("QUESTION 6: CUSTOMER SEGMENTATION")
print("="*50)

median_spending = cleaned_dataset['total_spending'].median()
median_purchases = cleaned_dataset['purchase_count'].median()

# Define segment conditions
conditions = [
    (cleaned_dataset['total_spending'] > median_spending) &
    (cleaned_dataset['purchase_count'] > median_purchases) &
    (cleaned_dataset['last_purchase_days'] <= 90) &
    (cleaned_dataset['satisfaction_score'] >= 4),

    (cleaned_dataset['total_spending'] > median_spending) &
    (cleaned_dataset['purchase_count'] > median_purchases) &
    (cleaned_dataset['last_purchase_days'] <= 180) &
    (cleaned_dataset['last_purchase_days'] > 90) &
    (cleaned_dataset['satisfaction_score'] >= 3),

    (cleaned_dataset['total_spending'] > median_spending) &
    (cleaned_dataset['last_purchase_days'] > 180) &
    (cleaned_dataset['purchase_count'] > 0)
]
def assign_segment(row):
    if (row['total_spending'] > median_spending) and (row['purchase_count'] > median_purchases) and (row['last_purchase_days'] <= 90) and (row['satisfaction_score'] >= 4):
        return 'Champions'
    elif (row['total_spending'] > median_spending) and (row['purchase_count'] > median_purchases) and (row['last_purchase_days'] <= 180) and (row['last_purchase_days'] > 90) and (row['satisfaction_score'] >= 3):
        return 'Loyal'
    elif (row['total_spending'] > median_spending) and (row['last_purchase_days'] > 180) and (row['purchase_count'] > 0):
        return 'At Risk'
    else:
        return 'Needs Attention'

cleaned_dataset['segment'] = cleaned_dataset.apply(assign_segment, axis=1)

# Segment summary
customer_count = cleaned_dataset.groupby('segment')['customer_id'].count()
avg_spending = cleaned_dataset.groupby('segment')['total_spending'].mean()
avg_purchases = cleaned_dataset.groupby('segment')['purchase_count'].mean()
avg_satisfaction = cleaned_dataset.groupby('segment')['satisfaction_score'].mean()
avg_recency = cleaned_dataset.groupby('segment')['last_purchase_days'].mean()

segment_summary = pd.DataFrame({
    'customer_count': customer_count,
    'avg_spending': avg_spending,
    'avg_purchases': avg_purchases,
    'avg_satisfaction': avg_satisfaction,
    'avg_recency': avg_recency
})
print("\nSegment Summary:")
display(segment_summary)

print("\n--- Segment Details & Actions ---")
print("""
۱. قهرمانان:
   - ویژگی‌ها: هزینه‌ی بالا، خریدهای مکرر، فعالیت اخیر، رضایت بالا.
   - اقدام: پاداش با مزایای VIP، دسترسی زودهنگام به محصولات جدید، و هدایای انحصاری.

۲. وفاداران:
   - ویژگی‌ها: هزینه‌ی بالا، خریدهای مکرر، فعالیت نسبتاً اخیر.
   - اقدام: تعامل با توصیه‌های شخصی‌سازی‌شده و برنامه‌های امتیاز وفاداری.

۳. در معرض ریسک:
   - ویژگی‌ها: هزینه‌ی بالا اما دوره‌های طولانی عدم فعالیت.
   - اقدام: راه‌اندازی کمپین‌های بازگردانی با مشوق‌های قوی.

۴. نیازمند توجه:
   - ویژگی‌ها: هزینه‌ی پایین، رضایت پایین، یا عدم فعالیت طولانی.
   - اقدام: ارسال نظرسنجی برای شناسایی نقاط ضعف و در نظر گرفتن پیشنهادات عمومی بازفعال‌سازی.
""")

# ==============================
# QUESTION 7 – Pareto Analysis (Revenue Concentration)
# ==============================
print("\n" + "="*50)
print("QUESTION 7: PARETO ANALYSIS (REVENUE CONCENTRATION)")
print("="*50)

df_pareto = cleaned_dataset.sort_values('total_spending', ascending=False).copy()

df_pareto = df_pareto.reset_index(drop=True)

total_customers = len(df_pareto)

df_pareto['cumulative_spending'] = df_pareto['total_spending'].cumsum()

total_revenue = df_pareto['total_spending'].sum()

df_pareto['cumulative_percent'] = (df_pareto['cumulative_spending'] / total_revenue) * 100

df_pareto['customer_percent'] = ((df_pareto.index + 1) / total_customers) * 100

top_20_count = int(0.2 * total_customers)

revenue_share_top20 = df_pareto.iloc[top_20_count - 1]['cumulative_percent']

print(f"Top 20% of customers generate {revenue_share_top20:.1f}% of total revenue.")

plt.figure(figsize=(10, 6))

plt.bar(df_pareto['customer_percent'],
        df_pareto['total_spending'],
        width=2, alpha=0.7, label='Spending per Customer')

plt.plot(df_pareto['customer_percent'],
         df_pareto['cumulative_percent'],
         color='red', marker='o', label='Cumulative %')

plt.axhline(y=80, color='blue', linestyle='--', label='80% Line')

plt.title('Pareto Analysis: Revenue Contribution by Customer', fontsize=14)
plt.xlabel('Customer Percentile (0% = poorest, 100% = richest)')
plt.ylabel('Total Spending / Cumulative %')

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n--- Interpretation ---")
print(f"""
داده‌ها الگوی ۸۰/۲۰ نسبتاً متعادلی را نشان می‌دهند (بالای ۲۰٪ مشتریان {revenue_share_top20:.1f}٪ از درآمد را تولید می‌کنند).
ریسک: اتکای بیش‌ازحد به پایگاه مشتریان کوچک. اگر این مشتریان کلیدی از دست بروند،
کسب‌وکار ممکن است با کاهش قابل‌توجه درآمد مواجه شود. تنوع‌بخشی توصیه می‌شود.
""")

# ==============================
# QUESTION 8 – CEO Dashboard (3 Charts)
# ==============================
print("\n" + "="*50)
print("QUESTION 8: CEO DASHBOARD - TOP 3 CHARTS")
print("="*50)

# ---------- Chart 1: Revenue by City (Bar Chart) ----------
# Step 1: Calculate total revenue for each city
city_revenue = cleaned_dataset.groupby('city')['total_spending'].sum().sort_values(ascending=False)

# Step 2: Draw a simple bar chart
plt.figure(figsize=(10,6))
plt.bar(city_revenue.index, city_revenue.values, color='skyblue')
plt.title('Total Revenue by City', fontsize=14)
plt.xlabel('City')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)  # Rotate city names so they don't overlap
plt.grid(axis='y', alpha=0.3)  # Light horizontal grid lines
plt.show()

# CEO insight for Chart 1
print("\nنکته کلیدی برای مدیرعامل در مورد نمودار ۱ (درآمد به تفکیک شهر):")
print("- این نمودار نشان می‌دهد که کدام شهرها بیشترین درآمد را ایجاد می‌کنند.")
print("- تصمیم: بودجه‌ی تبلیغات را در شهر برتر سرمایه‌گذاری کنید.")

# ---------- Chart 2: Scatter Plot (Spending vs. Last Purchase Days) ----------
# Step 1: Create a figure
plt.figure(figsize=(10,6))

# Step 2: Plot each customer segment with a different color
# (We already created the 'segment' column in Question 6)
segments = cleaned_dataset['segment'].unique()
colors = ['green', 'blue', 'orange', 'red']  # One color for each segment

for i, seg in enumerate(segments):
    # Get rows belonging to this segment
    subset = cleaned_dataset[cleaned_dataset['segment'] == seg]
    # Draw scatter points for this segment
    plt.scatter(subset['last_purchase_days'],
                subset['total_spending'],
                label=seg,
                alpha=0.7,
                color=colors[i % len(colors)])

# Step 3: Add labels and title
plt.title('Customer Value vs. Recency (Segmented)', fontsize=14)
plt.xlabel('Days Since Last Purchase')
plt.ylabel('Total Spending')
plt.legend(title='Segment')
plt.grid(True, alpha=0.3)
plt.show()

# CEO insight for Chart 2
print("""
نکته کلیدی برای مدیرعامل در مورد نمودار ۲ (نمودار پراکندگی):
- این نمودار مشتریان باارزش را نشان می‌دهد و اینکه هر کدام چقدر اخیراً خرید کرده‌اند.
- به‌وضوح مشتریان 'در معرض ریسک' را مشخص می‌کند (خرج بالا اما خرید قدیمی).
- تصمیم: کمپین‌های بازگردانی را بر روی ناحیه‌ی بالا-راست متمرکز کنید.
""")
# ---------- Chart 3: Boxplot (Spending Comparison by Discount) ----------
# Step 1: Use pandas built-in boxplot (very easy for beginners)
plt.figure(figsize=(8,6))
cleaned_dataset.boxplot(column='total_spending', by='discount_used', grid=True)

# Step 2: Add titles and clean up
plt.title('Total Spending: Discount vs. Non-Discount Customers', fontsize=14)
plt.suptitle('')  # Remove the automatic extra title
plt.xlabel('Used Discount')
plt.ylabel('Total Spending')
plt.show()

# CEO insight for Chart 3
print("""
نکته کلیدی برای مدیرعامل در مورد نمودار ۳ (نمودار جعبه‌ای):
- این نمودار نشان می‌دهد که مشتریانی که از تخفیف استفاده کرده‌اند، تمایل به خرج کردن بیشتری دارند.
- اما تنوع زیادی وجود دارد، بنابراین نمی‌توانیم مطمئن باشیم که تخفیف‌ها باعث افزایش فروش می‌شوند.
- تصمیم: قبل از گسترش برنامه‌های تخفیف، یک آزمایش کنترل‌شده (تست A/B) اجرا کنید.
""")

# ==============================
# QUESTION 9 – Three Data-Driven Proposals
# ==============================
print("\n" + "="*50)
print("QUESTION 9: THREE DATA-DRIVEN PROPOSALS")
print("="*50)

print("""
پیشنهاد ۱: تمرکز بودجه‌ی تبلیغات بر روی شهر با عملکرد برتر
   شواهد: تحلیل شهرها نشان می‌دهد که [شهر منتخب] بیشترین درآمد را تولید می‌کند و دارای پایگاه مشتریان قوی است.
   اقدام: اختصاص ۶۰٪ از بودجه‌ی تبلیغات به این شهر با پیام‌های متناسب.
   شاخص کلیدی (KPI): افزایش درآمد از این شهر حداقل به میزان ۱۵٪ طی ۲ ماه.

پیشنهاد ۲: کمپین بازگرداندن مشتریان در معرض ریسک
   شواهد: تعداد [len(at_risk)] مشتری باارزش که بیش از ۱۸۰ روز غیرفعال بوده‌اند، شناسایی شدند.
   اقدام: ارسال یک کد تخفیف ۲۰٪ شخصی‌سازی‌شده با تاریخ انقضای ۷ روزه از طریق ایمیل و پیامک.
   شاخص کلیدی (KPI): دستیابی به نرخ بازخرید ۱۵٪ و بازیابی ۱۰٪ از درآمد از دست‌رفته از این گروه طی ۳۰ روز.

پیشنهاد ۳: تقویت برنامه‌ی وفاداری برای بخش‌های قهرمانان و وفاداران
   شواهد: مشتریان بخش قهرمانان و وفاداران (به‌صورت ترکیبی) سهم عمده‌ای از کل درآمد را تشکیل می‌دهند.
   اقدام: راه‌اندازی یک سیستم امتیازدهی وفاداری پلکانی که در آن امتیازها به تخفیف‌های انحصاری یا دسترسی زودهنگام تبدیل می‌شوند.
   شاخص کلیدی (KPI): افزایش میانگین دفعات خرید این گروه‌ها به میزان ۱۰٪ طی ۳ ماه.
""")


# ==============================
# QUESTION 10 – Dataset Limitations
# ==============================
print("\n" + "="*50)
print("QUESTION 10: DATASET LIMITATIONS")
print("="*50)

print("""
محدودیت‌ها:
۱. عدم وجود سود به ازای هر سفارش:
   - ما فقط درآمد داریم، نه سود. درآمد بالا به معنای سود بالا نیست.

۲. عدم وجود هزینه‌ی حمل‌ونقل یا هزینه‌های عملیاتی:
   - این موارد به‌طور مستقیم بر سود خالص تأثیر می‌گذارند و برای ارزیابی عملکرد شهرها حیاتی هستند.

۳. عدم وجود اطلاعات دسته‌بندی محصولات:
   - نمی‌توانیم تشخیص دهیم که کدام دسته‌بندی‌ها باعث فروش یا بازگشت کالا می‌شوند، که این امر استراتژی محصول را محدود می‌کند.

۴. عدم وجود منبع جذب مشتری:
   - بدون دانستن اینکه مشتریان چگونه فروشگاه را پیدا کرده‌اند، نمی‌توانیم کانال‌های بازاریابی را بهینه کنیم.

۵. عدم وجود تاریخ دقیق سفارش‌ها:
   - فقط تعداد روزهای گذشته از آخرین خرید ارائه شده است. نمی‌توانیم روندهای فصلی یا بازه‌های خرید را تحلیل کنیم.

ستون‌های پیشنهادی برای بهبود تحلیل:
- سود به ازای هر سفارش
- هزینه‌ی حمل‌ونقل
- دسته‌بندی محصول
- تاریخ دقیق سفارش‌ها (یا زمان هر خرید)
- منبع جذب مشتری (مثلاً رسانه‌های اجتماعی، معرفی، ایمیل)
- هزینه‌ی تبلیغات به ازای هر مشتری یا هر شهر
""")
print("\n" + "="*50)
print("END OF ANALYSIS")
print("="*50)


```
