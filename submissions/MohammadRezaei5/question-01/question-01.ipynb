```python
import pandas as pd

# ۱. بارگذاری فایل داده‌ها
df = pd.read_excel('Cleaned_Customers_Dataset.xlsx')

# ۲. محاسبات شاخص‌های خواسته‌شده
total_customers = df['customer_id'].nunique()
total_purchases = df['purchase_count'].sum()
total_revenue = df['total_spending'].sum()
avg_spending_per_customer = df['total_spending'].mean()
avg_order_value = total_revenue / total_purchases  # vs df['avg_order_value'].mean()
avg_satisfaction = df['satisfaction_score'].mean()
total_returned = df['returned_items'].sum()
return_rate = (total_returned / total_purchases) * 100

# خروجی شاخص‌ها
print(f"تعداد کل مشتریان: {total_customers}")
print(f"تعداد کل خریدها: {total_purchases}")
print(f"درآمد کل: {total_revenue:,.2f}")
print(f"متوسط هزینه هر مشتری: {avg_spending_per_customer:,.2f}")
print(f"متوسط مبلغ هر خرید: {avg_order_value:,.2f}")
print(f"متوسط رضایت مشتریان: {avg_satisfaction:.2f} از ۵")
print(f"نرخ بازگشت کالا: {return_rate:.2f}%")
```
